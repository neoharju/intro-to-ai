# Metric

Before solving any task, it is important to understand how your solution will be evaluated. In this case, we use the standard classification metric **ROC-AUC**. It can be calculated using only the predicted probabilities and the true classes, without choosing a specific classification threshold. In addition, it works well even if the classes in the dataset are highly imbalanced (for example, when one class has tens of times more examples than the other). This makes it very convenient for competitions.

It is easy to calculate:



```python
from sklearn.metrics import roc_auc_score

y_true = [
    0,
    1,
    1,
    0,
    1
]

y_predictions = [
    0.1,
    0.9,
    0.4,
    0.6,
    0.61
]

roc_auc_score(y_true, y_predictions)
```
